# Data Preprocessing

```python
print("Dataset Shape:", df.shape)
```

```python
df.columns = df.columns.str.strip()
```

```python
df['Total Family Members'] = pd.to_numeric(df['Total Family Members'], errors='coerce')
```

```python
df['Village Name'] = df['Village Name'].str.strip().str.lower().str.replace(r'\s+', ' ', regex=True)
df['Village Name'].unique()
```

```python
df['Village Name'] = df['Village Name'].str.title()
df['Village Name'].value_counts()
```

```python
invalid_entries = ['Jarnal Singh', 'Rasid Mahomad','Sukhwinder', 'Charanjit Singh','Village']
df = df[~df['Village Name'].isin(invalid_entries)]
```
